```python
# ============================================================
# STEP 6: FEW-SHOT PROMPTING — EXAMPLE 1
# ============================================================

# Few-shot prompting provides the model with several examples
# before asking it to solve a new problem.

# This example uses arithmetic word problems.
few_shot_prompt_1 = """
Example 1:
A book costs $40 and has a 25% discount.
Final price: $30

Example 2:
A shirt costs $60 and has a 10% discount.
Final price: $54

Now solve this problem:

A pair of shoes costs $120 and has a 20% discount.

Provide only the final price.
"""

# Send the few-shot prompt to Gemini through LangChain.
few_shot_response_1 = llm.invoke(few_shot_prompt_1)

# Display the prompt used in the experiment.
print("FEW-SHOT PROMPT — EXAMPLE 1")
print("-" * 50)
print(few_shot_prompt_1)

# Display only the generated response text.
print("\nMODEL RESPONSE")
print("-" * 50)
print(few_shot_response_1.text)
```

```python
# ============================================================
# STEP 7: FEW-SHOT PROMPTING — EXAMPLE 2
# ============================================================

# Few-shot prompting provides labeled examples first.
# The model then follows the same pattern for a new input.

# This example tests sentiment classification.
few_shot_prompt_2 = """
Example 1:
Text: The movie was amazing and I loved every scene.
Sentiment: Positive

Example 2:
Text: The service was terrible and the staff was rude.
Sentiment: Negative

Example 3:
Text: The product is okay, but nothing special.
Sentiment: Neutral

Now classify the following text:

Text: The laptop is fast, reliable, and works better than I expected.

Answer only with:
Positive
Negative
or
Neutral
"""

# Send the few-shot prompt to Gemini through LangChain.
few_shot_response_2 = llm.invoke(few_shot_prompt_2)

# Display the prompt used in the experiment.
print("FEW-SHOT PROMPT — EXAMPLE 2")
print("-" * 50)
print(few_shot_prompt_2)

# Display only the generated response text.
print("\nMODEL RESPONSE")
print("-" * 50)
print(few_shot_response_2.text)
```

```python
# ============================================================
# STEP 8: CHAIN-OF-THOUGHT PROMPTING — EXAMPLE 1
# ============================================================

# Chain-of-Thought prompting demonstrates how to solve a problem
# using intermediate reasoning steps before giving the final answer.
#
# The model first sees a solved example with reasoning and is then
# asked to follow the same reasoning pattern for a new problem.

chain_of_thought_prompt_1 = """
Example:

Question:
A store has 50 notebooks.
It sells 18 notebooks in the morning and 12 in the afternoon.
How many notebooks are left?

Reasoning:
Start with 50 notebooks.
After selling 18 notebooks: 50 - 18 = 32.
After selling another 12 notebooks: 32 - 12 = 20.
Therefore, 20 notebooks are left.

Now solve the following problem using the same step-by-step format:

Question:
A library has 120 books.
It lends 35 books on Monday and receives 15 returned books on Tuesday.
Then it lends another 20 books on Wednesday.

How many books are in the library now?

Show the reasoning steps and then give the final answer.
"""

# Send the Chain-of-Thought prompt to Gemini through LangChain.
chain_of_thought_response_1 = llm.invoke(chain_of_thought_prompt_1)

# Display the prompt used in the experiment.
print("CHAIN-OF-THOUGHT PROMPT — EXAMPLE 1")
print("-" * 50)
print(chain_of_thought_prompt_1)

# Display only the generated response text.
print("\nMODEL RESPONSE")
print("-" * 50)
print(chain_of_thought_response_1.text)
```

```python
# ============================================================
# STEP 9: CHAIN-OF-THOUGHT PROMPTING — EXAMPLE 2
# ============================================================

# Chain-of-Thought prompting provides a worked reasoning example.
# The model is then asked to solve a new problem using the same
# step-by-step reasoning structure.

# This example tests logical reasoning.
chain_of_thought_prompt_2 = """
Example:

Question:
All cats are mammals.
All mammals are animals.
Can we conclude that all cats are animals?

Reasoning:
Cats belong to the group of mammals.
Mammals belong to the group of animals.
Therefore, cats also belong to the group of animals.

Final answer:
Yes

Now solve the following problem using the same step-by-step format:

Question:
All programmers are problem solvers.
Some problem solvers are musicians.

Can we conclude that some programmers are musicians?

Show the reasoning steps and then give the final answer.
"""

# Send the Chain-of-Thought prompt to Gemini through LangChain.
chain_of_thought_response_2 = llm.invoke(chain_of_thought_prompt_2)

# Display the prompt used in the experiment.
print("CHAIN-OF-THOUGHT PROMPT — EXAMPLE 2")
print("-" * 50)
print(chain_of_thought_prompt_2)

# Display only the generated response text.
print("\nMODEL RESPONSE")
print("-" * 50)
print(chain_of_thought_response_2.text)
```

```python
# ============================================================
# STEP 12: META-PROMPTING — EXAMPLE 1
# ============================================================

# Meta-prompting asks the model to first determine
# an appropriate strategy for solving the problem.
# The model then applies that strategy to produce the answer.

# This example uses a multi-step arithmetic problem.
meta_prompt_1 = """
You are a problem-solving assistant.

First, identify the best strategy for solving the problem.
Briefly describe the strategy.

Then apply that strategy to solve the problem.

Problem:
A company has 240 employees.
25% of the employees work remotely.
Of the remaining employees, 30 are managers.

How many non-manager employees work on-site?

Provide:
1. Strategy
2. Solution
3. Final Answer
"""

# Send the meta-prompt to Gemini through LangChain.
meta_response_1 = llm.invoke(meta_prompt_1)

# Display the prompt used in the experiment.
print("META-PROMPTING — EXAMPLE 1")
print("-" * 50)
print(meta_prompt_1)

# Display only the generated response text.
print("\nMODEL RESPONSE")
print("-" * 50)
print(meta_response_1.text)
```

## Q1: Comparison and Findings

Six prompt engineering techniques were evaluated using two different examples for each technique. The experiments showed that the structure of the prompt had a noticeable effect on the detail and organization of the model's response.

| Technique | Observation |
|---|---|
| Zero-Shot | The model answered directly using only the task instructions. It produced correct and concise answers, but did not provide reasoning when it was not requested. |
| Few-Shot | Providing examples helped the model follow the expected answer format and task pattern. The model correctly followed the demonstrations for both arithmetic and sentiment classification. |
| Chain-of-Thought | Providing a worked reasoning example encouraged the model to follow a similar step-by-step reasoning structure when solving the new problem. |
| Zero-Shot Chain-of-Thought | The model generated step-by-step reasoning without requiring a worked example. Adding an instruction to reason through the problem produced more detailed explanations than standard zero-shot prompting. |
| Meta-Prompting | The model first identified an appropriate problem-solving strategy and then applied it. This produced a more structured response that separated strategy selection from problem solving. |
| Tree of Thoughts | The model explored multiple candidate solutions, evaluated their advantages and disadvantages, and selected a final solution. This produced the most detailed responses but also required substantially more output than the simpler prompting methods. |

### Overall Findings

For simple tasks, zero-shot prompting was the most straightforward and efficient method. When a particular output pattern or classification format needed to be demonstrated, few-shot prompting was helpful. For multi-step problems, Chain-of-Thought and Zero-Shot Chain-of-Thought generated more explicit intermediate reasoning. By asking the model to choose a strategy prior to completing the task, meta-prompting introduced an extra planning step. By weighing and analyzing several options before deciding on a final response, Tree of Thoughts offered the most thorough investigation. All six methods yielded accurate final answers for the chosen tasks throughout these experiments. Nonetheless, there were notable differences in the quantity and organization of the generated reasoning. The extra reasoning required by advanced prompting techniques was not required for simple problems in order to get the right answer. Structured methods like Meta-Prompting and Tree of Thought can be used for more difficult reasoning or decision-making tasks.
